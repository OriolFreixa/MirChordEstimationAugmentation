## Test-only wrapper example

This cell shows how to run the new `test` mode in `ace_safe_trainer.py` without calling training first. Update the paths before running.

```python
from ace_safe_trainer import main as ace_model_wrapper

checkpoint_path = "training_jams_data/Checkpoints/last.ckpt"  # Replace with your checkpoint
train_data_path = "training_plus_augmentaiton/preprocessed_data"  # Replace if needed
test_data_path = "training_plus_augmentaiton/preprocessed_data"  # Replace with a different test folder if needed
vocab_path = "training_jams_data/chords_vocab.joblib"

ace_model_wrapper(
    model_name="conformer",
    run_name="code_test_eval",
    mode="test",
    checkpoint_path=checkpoint_path,
    params={
        "test.data_path": train_data_path,
        "test.test_data_path": test_data_path,
        "test.vocab_path": vocab_path,
        "test.accelerator": "gpu",
    },
)
```
